```python
import ipywidgets as widgets
from IPython.display import display
import random

btns=widgets.ToggleButtons(
    options = ['rock','paper','scissors'],
    description='Choose:'
)
out=widgets.Label()
def play(change):
  comp = random.choice(['rock','paper','scissors'])
  user = btns.value
  if user == comp:
      out.value = f"Tie! Both chose {comp}"
  elif(user=='rock' and comp=='scissors') or (user=='paper' and comp=='rock') or (user=='scissors' and comp=='paper'):
    out.value = f"You win! {user} Computer chose {comp}"
  else:
    out.value = f"You lose! {user} Computer chose {comp}"
btns.observe(play,names='value')
display(btns,out)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define common x values for continuous signals
x_values_continuous = np.linspace(-2 * np.pi, 2 * np.pi, 200)
x_values_discrete = np.arange(-2 * np.pi, 2 * np.pi, 0.5)
```

```python
# A: Exponential Signal
plt.figure(figsize=(10, 5))
plt.plot(x_values_continuous, np.exp(x_values_continuous), label='e^x', color='blue')
plt.title('Exponential Signal')
plt.xlabel('X-axis')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
G = 6      # Green duration (seconds)
Y = 2      # Yellow duration (seconds)
cycle_time = 3 * (G + Y)  # Total cycle time for all 3 roads
t = np.arange(0, 3 * cycle_time, 0.1)  # Show 3 cycles for clarity

# Initialize all signals with red (1) by default
signals = {
    'Road1': {'R': np.ones_like(t), 'G': np.zeros_like(t), 'Y': np.zeros_like(t)},
    'Road2': {'R': np.ones_like(t), 'G': np.zeros_like(t), 'Y': np.zeros_like(t)},
    'Road3': {'R': np.ones_like(t), 'G': np.zeros_like(t), 'Y': np.zeros_like(t)}
}

# Phase sequence: Road3 → Road1 → Road2
for i, time in enumerate(t):
    phase_time = time % cycle_time

    if phase_time < G:                    # Road 3 green
        signals['Road3']['R'][i] = 0
        signals['Road3']['G'][i] = 1
    elif phase_time < G + Y:              # Road 3 yellow
        signals['Road3']['R'][i] = 0
        signals['Road3']['Y'][i] = 1
    elif phase_time < 2*G + Y:            # Road 1 green
        signals['Road1']['R'][i] = 0
        signals['Road1']['G'][i] = 1
    elif phase_time < 2*(G + Y):          # Road 1 yellow
        signals['Road1']['R'][i] = 0
        signals['Road1']['Y'][i] = 1
    elif phase_time < 3*G + 2*Y:          # Road 2 green
        signals['Road2']['R'][i] = 0
        signals['Road2']['G'][i] = 1
    else:                                 # Road 2 yellow
        signals['Road2']['R'][i] = 0
        signals['Road2']['Y'][i] = 1

# Plot with better visualization
plt.figure(figsize=(14, 8))

# Colors for signals
signal_colors = {'R': 'red', 'G': 'green', 'Y': 'yellow'}

# Plot each signal
y_offset = 0
all_signals = []
all_labels = []

for road_idx, (road_name, road_signals) in enumerate(signals.items()):
    for signal_idx, (signal_name, signal_values) in enumerate(road_signals.items()):
        plt.fill_between(t,
                        y_offset,
                        y_offset + signal_values,
                        color=signal_colors[signal_name],
                        alpha=0.6,
                        edgecolor='black',
                        linewidth=0.5)

        # Add a line to separate signals
        plt.axhline(y=y_offset, color='gray', linestyle='-', linewidth=0.5)

        all_labels.append(f'{road_name} {signal_name}')
        y_offset += 1

# Add vertical lines for phase transitions
for cycle in range(4):  # Show transitions for multiple cycles
    for road_transition in range(3):
        transition_time = cycle * cycle_time + road_transition * (G + Y)
        plt.axvline(x=transition_time, color='blue', linestyle='--', alpha=0.3, linewidth=0.5)
        plt.axvline(x=transition_time + G, color='orange', linestyle='--', alpha=0.3, linewidth=0.5)

# Customize the plot
plt.yticks(np.arange(0.5, len(all_labels), 1), all_labels)
plt.xlabel('Time (seconds)', fontsize=12)
plt.ylabel('Traffic Signals', fontsize=12)
plt.title('Three-Road Traffic Signal Timing Diagram\n(Green: 6s, Yellow: 2s, Cycle: 24s)', fontsize=14, fontweight='bold')

# Add grid and limits
plt.grid(True, alpha=0.3)
plt.ylim(0, len(all_labels))
plt.xlim(0, max(t))

# Add legend for signal colors
legend_elements = [plt.Line2D([0], [0], color='red', lw=4, label='Red'),
                   plt.Line2D([0], [0], color='green', lw=4, label='Green'),
                   plt.Line2D([0], [0], color='yellow', lw=4, label='Yellow')]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# Print timing information
print("=== Traffic Signal Timing Information ===")
print(f"Green duration: {G} seconds")
print(f"Yellow duration: {Y} seconds")
print(f"Total cycle time: {cycle_time} seconds")
print(f"\nPhase sequence:")
print(f"1. Road 3: Green ({G}s) → Yellow ({Y}s)")
print(f"2. Road 1: Green ({G}s) → Yellow ({Y}s)")
print(f"3. Road 2: Green ({G}s) → Yellow ({Y}s)")
```
